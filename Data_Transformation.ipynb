**Q1. Handle Missing Values (Basic Cleaning)**

Fill missing values using the following:






*   Age → Median

*   Salary → Mean
*   Hours_Worked_Per_Week → Median


*   Performance_Score → Mean


Display the dataset after handling missing values

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df["Hours_Worked_Per_Week"] = df["Hours_Worked_Per_Week"].fillna(
    df["Hours_Worked_Per_Week"].median()
)

df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)
```

**Q7. Build Preprocessing Pipeline**

Create a pipeline that:


*   Applies encoding to categorical columns
*   Applies scaling to numerical columns






Use:


*   ColumnTransformer
*   Pipeline






```python
print(df.columns.tolist())
```

```python
categorical_columns = [
    "Gender",
    "Department",
    "Work_Mode",
    "Location"
]

numerical_columns = [
    "Age",
    "Salary",
    "Hours_Worked_Per_Week",
    "Performance_Score",
    "Projects_Completed"
]
```

```python
categorical_pipeline = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

numerical_pipeline = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            categorical_pipeline,
            categorical_columns
        ),
        (
            "numerical",
            numerical_pipeline,
            numerical_columns
        )
    ]
)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Categorical columns
categorical_columns = [
    "Gender",
    "Department",
    "Work_Mode",
    "Location"
]

# Numerical columns
numerical_columns = [
    "Age",
    "Salary",
    "Hours_Worked_Per_Week",
    "Performance_Score",
    "Projects_Completed"
]

# Categorical pipeline
categorical_pipeline = Pipeline(steps=[
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

# Numerical pipeline
numerical_pipeline = Pipeline(steps=[
    ("scaler", StandardScaler())
])

# Combine both pipelines
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", categorical_pipeline, categorical_columns),
        ("numerical", numerical_pipeline, numerical_columns)
    ]
)
```

**Q8. Apply Pipeline**

Apply the pipeline on the dataset.

Display:


*   Transformed dataset
*   Shape of final dataset






```python
# Apply the preprocessing pipeline
X_transformed = preprocessor.fit_transform(
    df_clean[categorical_columns + numerical_columns]
)

# Get the names of transformed columns
feature_names = preprocessor.get_feature_names_out()

# Convert transformed data into a DataFrame
transformed_df = pd.DataFrame(
    X_transformed,
    columns=feature_names
)

# Display transformed dataset
display(transformed_df.head())

# Display final shape
print("Final Shape:", transformed_df.shape)
```

**Q9. Conceptual Question**

Why is scaling important in Python?

**Answer**

Scaling is important in machine learning because different numerical features can have very different ranges. For example, Salary may have values in thousands while Performance Score may have values between 1 and 10. Features with larger numerical values can disproportionately influence some machine learning algorithms. Scaling brings features to comparable ranges and can improve model performance, stability, and convergence.

Why do we convert categorical data into numerical form?

**Answer**

Most machine learning algorithms work with numerical values rather than text categories. Therefore, categorical data such as Gender, Department, Work Mode, and Location must be converted into numerical representations before being provided to a machine learning model.
